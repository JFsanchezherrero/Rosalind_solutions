## Init4: Conditions and loops

In the following exercise, we need to compute the sum of all odd integers from a through b, inclusively.

```python
integers2check="100 200"
## split string, convert to int and map to a list
list2get = list(map(int, integers2check.split(" ")))
list2get
```

```python
total_sum =0
for i in range(list2get[0], list2get[1]+1):
    if i % 2 == 1:
        total_sum += i

print(total_sum)
```

A more pythonic way using lamba and filter

```python
total_sum = sum(filter(lambda x: x % 2, range(list2get[0], list2get[1] + 1)))
print(total_sum)
```

## Init6: Dictionaries

Finally, we came to the last initial exercise. See details [here](https://rosalind.info/problems/ini6/)

```python
string_to_test = "We tried list and we tried dicts also we tried Zen"
string_to_test.split(" ")
```

```python
dict_text={}
for item in string_to_test.split(" "):
    try:
        dict_text[item] +=1
    except:
         dict_text[item] =1
```

```python
dict_text
```

```python
from collections import defaultdict

dict_text = defaultdict(int)

for word in string_to_test.split():
    dict_text[word] += 1

print(dict_text)
```

```python
for x in dict_text.keys():
    print(x, str(dict_text[x]), sep=" ")
```
