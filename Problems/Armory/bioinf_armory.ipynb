The very first basic problem is to count DNA nucleotides. Find additional information in the rosalind website link [here](https://rosalind.info/problems/ini/).

Find previous information and example of solutions in the Stronghold solutions interactive python notebook here [`../Stronghold/stronghold_solutions.ipynb`](../Stronghold/stronghold_solutions.ipynb)

```python
dna_string = "AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC"
```

Now, try to be more ellaborate and produce a more programmatic solution

A very pythonic solution...

```python
## Solution 1: Using a python string
nucs = ["A", "C", "G", "T"]
counts = [dna_string.count(nuc) for nuc in nucs]
final_string = " ".join(map(str, counts))
print(final_string)
```

```python
## solution 2: Using biopython seq
from Bio.Seq import Seq
my_seq = Seq(dna_string)

nuc_list = [my_seq.count(nuc) for nuc in nucs]
final_string = " ".join(map(str, nuc_list))
print(final_string)
```

## GenBank Introduction

The following exercise is about producing a NCBI Genbank search using python. See details [here](https://rosalind.info/problems/gbk/).


For details and examples for how Biopython works, take a look at the [Example_Biopython.ipynb](https://github.com/JFsanchezherrero/Python_course/blob/main/practical_exercises/biopython/Example_Biopython.ipynb) available within my Python_course [here](https://github.com/JFsanchezherrero/Python_course/tree/main).

The goal of this exercise is to retrieve the number of nucleotide genbank entries for a given genus name published between the dates specified in YYYY/M/D format.

```python
# as an example
information2use = '''Anthoxanthum
2003/7/25
2005/12/27'''
```

```python
info_list = information2use.split("\n")
```

Find details of all limits that can be applied in Genbank searches [here](https://www.ncbi.nlm.nih.gov/nuccore/limits)

### Filter by genus

First, create a simple search using just the genus name

```python
string2search = info_list[0] + "[ORGN]"

print(string2search)
```

```python
# load Biopython module Entrez
from Bio import Entrez
Entrez.email = 'JoseFrancisco.Sanchez@uab.cat'

#Connect GenBank:
handle = Entrez.esearch(db="nucleotide", term=string2search)
## provide the term of interest using appropiate tags [ORGN], [TITL]

# Save results (XML format as default)
# and close connection
results = Entrez.read(handle)
handle.close()
```

### Filter by genus and date 

For organism we used the key tag [ORGN]; for date, we have to use [PDAT] that accounts for publication date. See an example for our previous case.

```python
string2search = info_list[0] + "[ORGN] AND (\"" + info_list[1] + "\"[PDAT]:\"" + info_list[2] + "\"[PDAT])"

print(string2search)
```

```python
# load Biopython module Entrez
from Bio import Entrez
Entrez.email = 'JoseFrancisco.Sanchez@uab.cat'

#Connect GenBank:
handle = Entrez.esearch(db="nucleotide", term=string2search)
## provide the term of interest using appropiate tags [ORGN], [TITL]

# Save results (XML format as default)
# and close connection
results = Entrez.read(handle)
handle.close()
```

This list of items is a list of entries in Genbank database. As an example, the first entry (57283843) is https://www.ncbi.nlm.nih.gov/nuccore/57283843
